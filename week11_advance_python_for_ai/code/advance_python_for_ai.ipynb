```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        result = func()
        print("Something is happening after the function is called.")
        return result
    return wrapper

def greet():
    return "HELLO...!"

deco = my_decorator(greet)

print(deco())
```

```python
def line_deco(func):
    '''This is a decoretor that adds a line of equal signs before and after the function call.'''
    def wrapper():
        print("="*20)
        result = func()
        print(result)
        print("="*20)

    return wrapper
    
```

```python
def greet():
    return "How are you?"
```

```python
greet()
```

```python
@line_deco
def greet():
    return "How are you?"
```

```python
greet()
```

```python
def rani():
    return "I live my husband."
```

```python
@line_deco
def rani():
    return "I love my husband."
```

```python
rani()
```

```python
import functools


def time(func):
    '''This is a decorator that calculates the time taken to execute a ffunction.'''
    
    @functools.wraps(func)  # preserve original version
    def wrapper(*args,**kwargs):
        import time
        start_time = time.time()
        result = func(*args,**kwargs)
        end_time = time.time()
        print(f"Time taken to exicute function {func.__name__}: {end_time - start_time} seconds")
        print(f"The function {func.__name__} use for {func.__doc__}")
        return result
    return wrapper
```

```python
@time
def addition(a, b):
    """This is a function that adds two numbers."""
    return a + b
```

```python
addition(1,2)
```
