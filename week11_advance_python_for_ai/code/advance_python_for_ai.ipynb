### Concept 2.5 — Iterables, Iterators & Generators (dono mein farak)
Technical definitions:

- Iterable: An Iterable is any Python object (like a list, tuple, or string) that can be looped over in a for loop. It contains data in memory but does not track its own iteration state or current position. Passing an Iterable to iter(x) creates and returns a new Iterator object.
- Iterator: An Iterator is a stateful stream object that remembers its current position during iteration. It fetches the next item one-at-a-time whenever next(iterator) is called. When no elements remain, calling next() raises a StopIteration exception.
- Generator: A Generator is a special type of function that uses yield to return values lazily one at a time. Calling a generator function automatically returns a specialized Iterator object without storing all items in memory at once. It pauses execution at yield and resumes seamlessly when the next item is requested, making it ideal for LLM token streaming.

### Concept 2 — decorator ka basic dhaancha
Technical definition — Decorator: A decorator is a function that takes another function (or class) and returns a modified version, wrapping it with extra behaviour without changing its original code. The @decorator syntax applies it to the function defined beneath it.

```python
def my_decorator(func):
    def wrapper():
        result = func()
        return result
    return wrapper

def greet():
    return "HELLO...!"

deco = my_decorator(greet)

deco()
```

```python
print(deco())
```

```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        result = func()
        print("Something is happening after the function is called.")
        return result
    return wrapper

def greet():
    return "HELLO...!"

deco = my_decorator(greet)

print(deco())
```

### Concept 3 — @ syntax (sundar shortcut)


```python
@my_decorator

def greet():
    return "HELLO...!"
```

```python
greet()
```

```python
def line_deco(func):
    '''This is a decoretor that adds a line of equal signs before and after the function call.'''
    def wrapper():
        print("="*20)
        result = func()
        print(result)
        print("="*20)

    return wrapper
    
```

```python
def greet():
    return "How are you?"
```

```python
greet()
```

```python
@line_deco
def greet():
    return "How are you?"
```

```python
greet()
```

```python
def rani():
    return "I live my husband."
```

```python
@line_deco
def rani():
    return "I love my husband."
```

### Concept 4 — arguments wale functions ko wrap karna (*args, **kwargs)


```python
def addition(a, b):
    return a + b
```

```python
addition(1,2)
```

```python
import time
```

```python
import functools


def time(func):
    '''This is a decorator that calculates the time taken to execute a ffunction.'''
    
    @functools.wraps(func)  # preserve original version
    def wrapper(*args,**kwargs):
        import time
        start_time = time.time()
        result = func(*args,**kwargs)
        end_time = time.time()
        print(f"Time taken to exicute function {func.__name__}: {end_time - start_time} seconds")
        print(f"The function {func.__name__} use for {func.__doc__}")
        return result
    return wrapper
```

```python
@time
def addition(a, b):
    """This is a function that adds two numbers."""
    return a + b
```

```python
addition(1,2)
```

```python
addition.__name__
```

```python
addition.__doc__
```

```python
@time
def mult(a,b,c):
    """This is a function that multiply three numbers."""
    return a* b* c
```

```python
mult(1,2,3)
```

# CLASS 63 — Context Managers & collections

### Concept 1 — context manager kya karta hai
 Technical definition — Context manager: A context manager is an object used with the with statement that runs setup code on entry and guaranteed cleanup code on exit, even if an error occurs. It is commonly created with the @contextmanager decorator, where code before yield is setup and code after it is cleanup.

### Concept 2 — contextmanager se apna banao

```python
from contextlib import contextmanager
import time
```

```python
@contextmanager
def timer(label):
    start_time = time.time()
    print(f"Starting {label}...")
    print("Starting time is", start_time)
    yield

    end_time = time.time()
    print(f"{label} took {end_time - start_time} seconds")

```

```python
with timer("my task"):
    total = sum(range(1000))
    print("Total sum is", total)
```

### Concept 3 — functools.lru_cache (results yaad rakho)
📖 Technical definition — lru_cache: lru_cache is a decorator that caches (memoizes) a function's return values keyed by its arguments. When the same arguments are seen again, the stored result is returned instantly instead of recomputing, avoiding repeated expensive work.

```python
from functools import lru_cache
```

```python
@lru_cache(maxsize=None)
def slow_sq(n):
    print(f"Calculating {n} square.......")
    return n ** 2
```

### Concept 6 — Protocol (first pass — duck typing)
📖 Technical definition — Protocol: A Protocol defines an interface by the methods and attributes an object must have, rather than by its class. Any object that provides those members satisfies the protocol (structural, "duck" typing), enabling flexible type checking.

```python
from typing import Protocol
```

```python
class Runnable(Protocol):
    def run(self) -> str:...

def exicute(tool: Runnable) -> str:
    return tool.run()
```

```python
class SearchTool:
    def run(self) -> str:
        return "Searching......."
```

```python
tool = SearchTool()

exicute(tool)
```

```python
class SearchToolWithoutRun:
    def ran(self) -> str:
        return "Searching......."
```

```python
bad_tool = SearchToolWithoutRun

exicute(bad_tool)
```
